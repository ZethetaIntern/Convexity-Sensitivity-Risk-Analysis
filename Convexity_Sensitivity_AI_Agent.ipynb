## Risk Contribution by Rating in Python

```python
# ==========================================
# RISK CONTRIBUTION BY CREDIT RATING
# ==========================================

rating_risk = df.groupby('CreditRating').agg(
    Duration_Contribution=(
        'ModifiedDuration',
        lambda x: (x * df.loc[x.index, 'PortfolioWeight']).sum()
    ),
    Convexity_Contribution=(
        'Convexity',
        lambda x: (x * df.loc[x.index, 'PortfolioWeight']).sum()
    ),
    DV01_Contribution=(
        'DV01_Per100Face',
        lambda x: (
            x * df.loc[x.index, 'Quantity']
        ).sum() / 100
    )
).reset_index()

rating_risk
```

## Yield Shock / Sensitivity Analysis

## Create the sensitivity table

```python
# Yield Shock Sensitivity Analysis

shocks_bps = [-200, -100, -50, 50, 100, 200]

portfolio_value = df['MarketValue_INR'].sum()

portfolio_duration = (
    df['ModifiedDuration'] * df['PortfolioWeight']
).sum()

portfolio_convexity = (
    df['Convexity'] * df['PortfolioWeight']
).sum()

sensitivity = pd.DataFrame({
    'Shock_bps': shocks_bps
})

# Convert bps to decimal
sensitivity['Shock'] = sensitivity['Shock_bps'] / 10000

# Duration effect
sensitivity['Duration_Effect_%'] = (
    -portfolio_duration * sensitivity['Shock'] * 100
)

# Convexity effect
sensitivity['Convexity_Effect_%'] = (
    0.5 * portfolio_convexity *
    sensitivity['Shock']**2 * 100
)

# Total estimated price change %
sensitivity['Estimated_Price_Change_%'] = (
    sensitivity['Duration_Effect_%'] +
    sensitivity['Convexity_Effect_%']
)

# Estimated portfolio P&L
sensitivity['Estimated_PnL_INR'] = (
    portfolio_value *
    sensitivity['Estimated_Price_Change_%'] / 100
)

# Estimated portfolio value after shock
sensitivity['Estimated_Portfolio_Value'] = (
    portfolio_value +
    sensitivity['Estimated_PnL_INR']
)

sensitivity
```

```python
sensitivity.to_csv(
    'yield_shock_sensitivity.csv',
    index=False
)
```

## Price Impact Analysis

### Calculate the exact price impact

```python
# Price Impact Analysis

price_impact = sensitivity[
    [
        'Shock_bps',
        'Duration_Effect_%',
        'Convexity_Effect_%',
        'Estimated_Price_Change_%',
        'Estimated_PnL_INR',
        'Estimated_Portfolio_Value'
    ]
].copy()

price_impact
```

### Identify the worst-case scenario

```python
worst_case = price_impact.loc[
    price_impact['Estimated_PnL_INR'].idxmin()
]

worst_case
```

## Identify the best-case scenario

```python
best_case = price_impact.loc[
    price_impact['Estimated_PnL_INR'].idxmax()
]

best_case
```

## final price-impact table

```python
price_impact.to_csv(
    'price_impact_analysis.csv',
    index=False
)
```

## Final Scenario Summary

### Create the final scenario summary

```python
# Final Scenario Summary

scenario_summary = sensitivity[
    [
        'Shock_bps',
        'Estimated_Price_Change_%',
        'Estimated_PnL_INR',
        'Estimated_Portfolio_Value'
    ]
].copy()

scenario_summary
```
